# Part B: Clean the data
* Find and remove duplicate rows.
* Standardize text columns so the same value is always written the same way.
* Handle missing ratings. You can drop them or leave them out of averages, but explain your choice in Markdown.
* Print the number of rows before and after cleaning.



```python
# 1. Deduplication
print(f"{"Remove Duplicates":-^60}")
print()

# Print Original number of rows
print("Original Data rows count:",len(df))

# Find duplicate rows
duplicates = df[df.duplicated()]

# Print number of duplicate rows
print("Duplicate rows count:",len(duplicates))

# Created a copy to prevent warning later when trying to change case of values in column later
df_cleaned = df.drop_duplicates().copy()

# Print number of rows after dropping duplicates
print("Data rows count after drop dupliates:",len(df_cleaned))
```

```python
# 2. Standardize Product name (convert case of product to lower)
print(f"{"Standardize Product Names":-^60}\n")

# Print unique product names
print("Unique products before cleaning:\n")
print(df_cleaned['product'].unique())
df_cleaned['product'] = df_cleaned['product'].str.lower()

print("\nUnique products AFTER cleaning:\n")
print(df_cleaned['product'].unique())
```

```python
#3. Trim spaces before and after store location, standardize case
print(f"{"Standardize Store Location Name":-^60}\n")

# Print unique product names
print("Unique Store Location before cleaning:\n")
print(df_cleaned['store_location'].unique())

df_cleaned['store_location'] = df_cleaned['store_location'].str.capitalize()
df_cleaned['store_location'] = df_cleaned['store_location'].str.strip()

print("Unique Store Location after cleaning:\n")
print(df_cleaned['store_location'].unique())
```

### C1. What are the total revenue, total number of orders and average order value?

```python
total_revenue = sum(df_cleaned['total_price'])
total_num_orders = len(df_cleaned['order_id'].unique())
avg_order_value = df_cleaned['total_price'].mean()

report_c1 = f"""
{'='*10} 1. SALES OVERVIEW {'='*10}
Total Revenue       : ${total_revenue:,}
Total Orders        : {total_num_orders}
Average order value : ${avg_order_value:.2f}
"""
print(report_c1)
```

###C2. What are the top 5 products by revenue?

```python
# Get a df from cleaned df, grouped by product, sums total price for each group
top5 = df_cleaned.groupby('product')['total_price'].sum().sort_values(ascending=False)

# Convert df head (only 5) to series
top5Series = (
    df_cleaned
    .groupby('product')['total_price']
    .sum()
    .sort_values(ascending=False)
    .head()
    )

# print the product and price
top5printable = '\n'.join(
    f"{index}. {prod}...... ${total_price}"
    for index,(prod,total_price) in enumerate(top5Series.items(),start=1)
)

report_c2 = f"""{'='*5} 2. TOP 5 PRODUCTS BY REVENUE {'='*5}
{top5printable}
"""
print(report_c2)
```

###C3. How much revenue does each category bring in, and which drink size is most popular?

```python
# Get a series from cleaned df, grouped by category, sums total price, sorted from highest to lowest
categoryWiseRevenue = (df_cleaned.groupby('category')['total_price'].sum().sort_values(ascending=False))

# Print category and its total revenue
category_print = "\n".join(
    f"{index}. {category}....... ${total_price:,.2f}"
    for index, (category,total_price) in enumerate(categoryWiseRevenue.items(), start=1)
)

# Get filtered dataframe that has a category of Tea or Coffee ( to get only drinks )
df_drinks = df_cleaned[df_cleaned['category'].isin(['Coffee', 'Tea'])]

# Get a series of data from df_drinks that counts the values in 'size' column. value_counts returns
# a series of count of each distinct row in the df
sizes = df_drinks.value_counts('size')

# For reporting:
report_c3 =  f"""{'='*8} 3. REVENUE PER CATEGORY {'='*8}
{category_print}

Most poular drink size : {sizes.index[0]}
Num of orders with {sizes.index[0]} size: {sizes.values[0]}
"""
print(report_c3)
```

###C4.	Which day of the week is the busiest (most orders)?

```python
# Get distinct value count for the colum day of week
days = df_cleaned.value_counts('day_of_week')
report_c4 = f'''{'='*8} 4. BUSIEST DAY OF WEEK {'='*8}
Busiest day of the week: {days.index[0]}
Num of orders on {days.index[0]}: {days.values[0]}
'''
print(report_c4)
```

 ### C5. What is the peak hour of the day?

```python
# slice the string in time column for first two char - so that we get the hour and then apply value_counts
hours = df_cleaned['time'].str[:2].value_counts()

report_c5 = f'''{'='*8} 5. PEAK HOUR OF A DAY {'='*8}
Peak hour of the day.      : {hours.index[0]}
Num of orders in this hour : {hours.values[0]}
'''
print(report_c5)
```

###C6. What are the revenue, order count and average rating for each store?


```python
store_wise_revenue = df_cleaned.groupby('store_location')['total_price'].sum()

store_wise_order_count = df_cleaned.groupby('store_location')['order_id'].count()

store_wise_avg_rating = df_cleaned.groupby('store_location')['rating'].mean()

# Concats the three data frames derived above
summary = pd.concat([store_wise_revenue, store_wise_order_count, store_wise_avg_rating], axis=1)

# resets the store location index - after using groupBy, store location is set as df's index
summary = summary.reset_index()

# Renames columns
summary.columns = ['Store','Revenue', 'Orders', 'Rating']

report_c6 = f'''{'='*8} 6. STOREWISE REPORT {'='*8}
{summary}
'''
print(report_c6)

# Recommendation based on Store Performance
summary_sorted = summary.sort_values('Revenue')
recommended_store = summary_sorted[summary_sorted['Rating']>4.0].iloc[0]
recommendation = f'''You can model your next store based on the {recommended_store['Store']} store.
This store have a top revenue and high customer rating.
'''
```

### C7. Who is Barista of the Quarter? Compare total sales and average rating, pick a winner and justify it.

```python
# Group by barista for total_revenue
barista_revenue = df_cleaned.groupby('barista')['total_price'].sum()

# Group by barista for average rating
barista_rating = df_cleaned.groupby('barista')['rating'].mean()

# Concatenate, sort by total revenu
barista_perf = pd.concat([barista_revenue,barista_rating],axis=1).reset_index().sort_values('total_price', ascending=False)

# Print barista perf report sorted by revenue generated
print("-------------Barista Performance Report------------\n")
print(barista_perf)
print()

# Find the barista who has highest revenue and rating > 4.0
barista_of_quarter = barista_perf[barista_perf['rating']>4.0].iloc[0]

# Print performer of the quarter
report_c7 = f'''{'='*5} 7. BARISTA OF THE QUARTER {'='*5}
Winner: {barista_of_quarter['barista']}
Reason: Highest revenue and Rating > 4.0
'''
print(report_c7)

```

### C8. How do customers pay (share of each payment method), and what share are new vs returning?

```python
# Share of payment method
payments = df_cleaned.groupby('payment_method')['order_id'].count().reset_index()

# Rename column
payments.columns = ['payment_method','num_orders']

# Find total num of orders
total_orders = len(df_cleaned['order_id'].unique())

# Add new column for share (percentage)
payments['share'] = 100 * payments['num_orders']/total_orders

# Share of new vs Returning
customerType = df_cleaned.groupby('customer_type')['order_id'].count().reset_index()

# Rename column
customerType.columns = ['customer_type','num_orders']

# Add new column for share (percentage)
customerType['share'] = 100 * customerType['num_orders']/total_orders

report_c8 = f'''{'='*15} 8. PATTERNS {'='*15}

{'-'*10} 8.1 PAYMENT PATTERNS {'-'*11}
{payments.sort_values('share',ascending=False).reset_index(drop=True)}

{'-'*10} 8.2 CUSTOMER PATTERN {'-'*11}
{customerType.sort_values('share',ascending=False).reset_index(drop=True)}
'''

print(report_c8)
```
